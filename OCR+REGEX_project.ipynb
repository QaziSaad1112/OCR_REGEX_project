```python
import pytesseract
from PIL import Image
import json

# load image
img = Image.open('/content/T.jpg.jpeg')

# apply ocr
text = pytesseract.image_to_string(img)
print(text)

# store ocr text in json
data = {
    "ocr_text": text
}

with open('/content/invoice_data.json', 'w', encoding='utf-8') as file:json.dump(data, file, indent=4, ensure_ascii=False)

```

```python
# read json file
with open('/content/invoice_data.json', 'r', encoding='utf-8') as file:
    data = json.load(file)

text = data["ocr_text"]
```

```python
import re
# get first shipper/exporter section
shiping_export = re.search(r'Shipper\s*/\s*Exporter(.*?)(?=Consignee)',text,re.S).group(1)

# extract name
name = re.findall(r'Name\s*\[([^\]]+)\]', shiping_export)

# extract country
country = re.findall(r'Country\s*\[([^\]]+)\]', shiping_export)
# extract address
address = re.findall(r'Address\s*\[([^\]]+)\]', shiping_export)

print("Name : ",name)
print("Country :",country)
print("Address :",address)

```

```python
import re
# extract invoice number
invoice_number = re.findall(r'Invoice Number\s*\[([^\]]+)\]',text)

# extract country of export
country_export = re.findall(r'Country of Export\s*\[([^\]]+)\]',text)



# extract country of destination
country_destination = re.findall(r'Country of Destination\s*\[([^\]]+)\]',text)

print("Invoice_number :",invoice_number)
print("Country of ecport :",country_export)
print("Country of destination :",country_destination)
```

```python
import re
# get last shipper/exporter section
bottom_section = text[text.rfind("Shipper / Exporter"):]

# extract name
bottom_name = re.findall(r'Name\s*\[([^\]]+)\]',bottom_section)
# extract job title
Job_titel = re.findall(r'Job Title\s*\[([^\]]+)\]',bottom_section)
# extract date
date = re.findall(r'Date\s*\[([^\]]+)\]',bottom_section)


print("Name :",bottom_name)
print("Job_titel :",Job_titel)
print("Date :",date)

```

```python
import re
# extract insurance cost
insurance_cost = re.findall(r'Insurance Costs\s+([\d.]+)',text)
# extract additional cost
additional_cost = re.findall(r'Additional Costs\s+([\d.]+)',text)
print("insurance cost :",insurance_cost)
print("Additional_cost :",additional_cost)
```
